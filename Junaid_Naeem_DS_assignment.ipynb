```python

# Create a safe working copy
df_raw = df.copy()
df_clean = df.copy()

print("Original dataset:", df_raw.shape)
print("Working copy:", df_clean.shape)
```

```python

# Check missing values in the working copy

missing_data = pd.DataFrame({
    "Missing Count": df_clean.isnull().sum(),
    "Missing Percentage": (df_clean.isnull().mean() * 100).round(2)
})

missing_data = missing_data[missing_data["Missing Count"] > 0]
missing_data = missing_data.sort_values(
    by="Missing Count", ascending=False
)

display(missing_data)
```

```python

# A. Fill missing numerical values with the median
# These columns have relatively few missing values.

for col in ["bathrooms", "bedrooms", "beds"]:
    median_value = df_clean[col].median()
    df_clean[col] = df_clean[col].fillna(median_value)

# B. Mark missing categorical information as Unknown
for col in ["neighbourhood", "zipcode",
            "host_has_profile_pic", "host_identity_verified"]:
    df_clean[col] = df_clean[col].fillna("Unknown")

print("Missing values handled for selected columns.")
```

```python

remaining = df_clean.isnull().sum()
display(remaining[remaining > 0].to_frame("Remaining Missing Values"))
```

### Handling Missing Values

Missing values were identified using `isnull().sum()` and their percentages were calculated. The missing values in `bathrooms`, `bedrooms`, and `beds` were filled using their median values because these numerical columns had relatively few missing records. Missing values in `neighbourhood`, `zipcode`, `host_has_profile_pic`, and `host_identity_verified` were replaced with `Unknown` to preserve the fact that the information was unavailable.

Missing review dates, review ratings, host response rates, and thumbnail URLs were retained for further inspection because their absence may have a meaningful explanation. The original dataset was preserved in `df_raw`, while cleaning was performed on `df_clean`.

```python

# Check complete duplicate rows
duplicate_rows = df_clean.duplicated().sum()

# Check duplicate listing IDs
duplicate_ids = df_clean["id"].duplicated().sum()

print("Complete duplicate rows:", duplicate_rows)
print("Duplicate listing IDs:", duplicate_ids)
```

### Duplicate Record Detection

The dataset was checked for complete duplicate rows using `df_clean.duplicated()` and for repeated listing IDs using `df_clean["id"].duplicated()`. Both checks returned zero duplicates. Therefore, no records were removed because there were no duplicate rows or repeated listing IDs.

```python

categorical_cols = [
    "property_type",
    "room_type",
    "bed_type",
    "cancellation_policy",
    "city",
    "host_has_profile_pic",
    "host_identity_verified",
    "instant_bookable"
]

for col in categorical_cols:
    print(f"\n--- {col} ---")
    print(df_clean[col].value_counts(dropna=False))
```

```python

# Standardize text values by removing extra spaces
for col in df_clean.select_dtypes(include="object").columns:
    df_clean[col] = df_clean[col].str.strip()

# Make host information easier to understand
binary_text_cols = [
    "host_has_profile_pic",
    "host_identity_verified"
]

for col in binary_text_cols:
    df_clean[col] = df_clean[col].replace({
        "t": "True",
        "f": "False"
    })

# Convert instant_bookable to Boolean values
df_clean["instant_bookable"] = df_clean["instant_bookable"].map({
    "t": True,
    "f": False
})

# Verify the changes
for col in [
    "host_has_profile_pic",
    "host_identity_verified",
    "instant_bookable"
]:
    print(f"\n{col}:")
    print(df_clean[col].value_counts(dropna=False))
```

```python
# Convert date columns into datetime format
date_cols = ["first_review", "host_since", "last_review"]

for col in date_cols:
    df_clean[col] = pd.to_datetime(df_clean[col], errors="coerce")

# Check the results
df_clean[date_cols].dtypes
```

```python
# Count missing values in each column
missing = df_clean.isnull().sum()

# Show only columns that still have missing values
missing = missing[missing > 0].sort_values(ascending=False)

print(missing)
```

```python
# Keep missing review dates and ratings as they are.
# They may indicate that a listing has no reviews or rating.

# Replace missing text information with "Unknown"
df_clean["host_response_rate"] = (
    df_clean["host_response_rate"].fillna("Unknown")
)

df_clean["thumbnail_url"] = (
    df_clean["thumbnail_url"].fillna("Unknown")
)

# Keep host_since as a datetime column.
# Missing dates remain NaT because the actual dates are unknown.

# Check remaining missing values
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
```

```python
# Select numerical columns
numeric_cols = [
    "log_price",
    "accommodates",
    "bathrooms",
    "bedrooms",
    "beds",
    "number_of_reviews",
    "review_scores_rating",
    "latitude",
    "longitude"
]

# Display summary statistics
df_clean[numeric_cols].describe().round(2)
```

```python
# Detect potential outliers using IQR
for col in [
    "log_price",
    "accommodates",
    "bathrooms",
    "bedrooms",
    "beds",
    "number_of_reviews",
    "review_scores_rating"
]:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = (
        (df_clean[col] < lower) |
        (df_clean[col] > upper)
    ).sum()

    print(f"{col}: {count} potential outliers")
```

```python
# Inspect the highest values in selected columns
for col in [
    "log_price",
    "accommodates",
    "bathrooms",
    "bedrooms",
    "beds",
    "number_of_reviews",
    "review_scores_rating"
]:
    print(f"\n--- {col} ---")
    print(df_clean[col].dropna().sort_values(ascending=False).head(10).to_list())
```

```python
# Create a separate copy for preprocessing
df_preprocessed = df_clean.copy()

# Remove the listing ID from potential model features
df_preprocessed = df_preprocessed.drop(columns=["id"])

print("Cleaned dataset shape:", df_clean.shape)
print("Preprocessing dataset shape:", df_preprocessed.shape)
```

```python
# Create useful features from dates
df_preprocessed["host_since_year"] = (
    df_preprocessed["host_since"].dt.year
)

df_preprocessed["first_review_year"] = (
    df_preprocessed["first_review"].dt.year
)

df_preprocessed["last_review_year"] = (
    df_preprocessed["last_review"].dt.year
)

# Count amenities listed for each property
def count_amenities(value):
    if pd.isna(value) or value == "Unknown" or value.strip() == "":
        return 0
    return len(value.strip("{}").split(","))

df_preprocessed["amenities_count"] = (
    df_preprocessed["amenities"].apply(count_amenities)
)

# Check the new features
print(df_preprocessed[
    [
        "host_since_year",
        "first_review_year",
        "last_review_year",
        "amenities_count"
    ].head()
])

print("\nNew dataset shape:", df_preprocessed.shape)
```

```python
# Display the first 5 rows of the new features
print(df_preprocessed[
    [
        "host_since_year",
        "first_review_year",
        "last_review_year",
        "amenities_count"
    ]
].head())

print("\nNew dataset shape:", df_preprocessed.shape)
```

```python
import numpy as np
```

```python
# Convert percentage text into numeric proportions
df_preprocessed["host_response_rate"] = (
    df_preprocessed["host_response_rate"]
    .replace("Unknown", np.nan)
    .str.rstrip("%")
    .astype(float) / 100
)

# Check the conversion
print(df_preprocessed["host_response_rate"].head(10))
print("\nMissing response rates:",
      df_preprocessed["host_response_rate"].isnull().sum())
```

```python
# Check missing values in the preprocessing dataset
missing_counts = df_preprocessed.isnull().sum()
print(missing_counts[missing_counts > 0].sort_values(ascending=False))
```

```python
# Record which values were missing before filling them
df_preprocessed["response_rate_missing"] = (
    df_preprocessed["host_response_rate"].isna().astype(int)
)

df_preprocessed["rating_missing"] = (
    df_preprocessed["review_scores_rating"].isna().astype(int)
)

df_preprocessed["first_review_missing"] = (
    df_preprocessed["first_review"].isna().astype(int)
)

df_preprocessed["last_review_missing"] = (
    df_preprocessed["last_review"].isna().astype(int)
)

df_preprocessed["host_since_missing"] = (
    df_preprocessed["host_since"].isna().astype(int)
)

# Fill missing numeric values with their medians
numeric_fill_cols = [
    "host_response_rate",
    "review_scores_rating",
    "host_since_year",
    "first_review_year",
    "last_review_year"
]

for col in numeric_fill_cols:
    df_preprocessed[col] = df_preprocessed[col].fillna(
        df_preprocessed[col].median()
    )

# Verify the selected columns
print(df_preprocessed[numeric_fill_cols].isnull().sum())
print("\nNew dataset shape:", df_preprocessed.shape)
```

### Categorical Encoding

One-hot encoding was selected for categorical features such as property type, room type, bed type, cancellation policy, and city. This creates separate binary columns for their categories without assigning artificial numerical rankings. Free-text fields and high-cardinality fields, such as listing names, descriptions, amenities, and URLs, were excluded from this basic encoding step because they require specialized processing.

```python
import pandas as pd

# Create a separate copy for encoding
df_encoded = df_preprocessed.copy()

# Select categorical columns suitable for one-hot encoding
categorical_cols = [
    "property_type",
    "room_type",
    "bed_type",
    "cancellation_policy",
    "city",
    "host_has_profile_pic",
    "host_identity_verified"
]

# Convert selected categorical columns into binary columns
df_encoded = pd.get_dummies(
    df_encoded,
    columns=categorical_cols,
    dtype=int
)

print("Encoded dataset shape:", df_encoded.shape)
print(df_encoded.head())
```

```python
from sklearn.preprocessing import StandardScaler

# Create another copy for scaling
df_scaled = df_encoded.copy()

# Select continuous numeric features to scale
scale_cols = [
    "log_price",
    "accommodates",
    "bathrooms",
    "bedrooms",
    "beds",
    "number_of_reviews",
    "review_scores_rating",
    "latitude",
    "longitude",
    "host_response_rate",
    "host_since_year",
    "first_review_year",
    "last_review_year",
    "amenities_count"
]

# Scale only the selected numeric columns
scaler = StandardScaler()
df_scaled[scale_cols] = scaler.fit_transform(df_scaled[scale_cols])

print("Scaled dataset shape:", df_scaled.shape)
print(df_scaled[scale_cols].describe().round(2))
```

### Numerical Scaling

Standardization was applied to selected numerical features using `StandardScaler`. The transformation places the features on comparable scales, with values centered around zero and scaled to approximately unit standard deviation. Binary indicators and one-hot encoded columns were not scaled. For future predictive modeling, the scaler must be fitted only on the training data and then applied to validation and test data to prevent data leakage.

```python
# Check skewness of selected numeric features
skew_cols = [
    "log_price",
    "accommodates",
    "number_of_reviews",
    "beds",
    "amenities_count"
]

print(df_preprocessed[skew_cols].skew().round(2))
```

```python
# Create a log-transformed review-count feature
df_preprocessed["number_of_reviews_log"] = np.log1p(
    df_preprocessed["number_of_reviews"]
)

# Compare the original and transformed distributions
print("Original skewness:",
      round(df_preprocessed["number_of_reviews"].skew(), 2))

print("Transformed skewness:",
      round(df_preprocessed["number_of_reviews_log"].skew(), 2))
```

### Skewness and Data Transformation

Skewness was examined for selected numerical features. A logarithmic transformation using `log1p()` was applied to the number of reviews to reduce the effect of very large review counts. The original feature was retained for comparison. The `log_price` feature was not transformed again because it was already log-transformed in the original dataset. Transformations should be evaluated before their use in a predictive model.

```python
# Final dataset checks
print("Original dataset:", df_raw.shape)
print("Cleaned dataset:", df_clean.shape)
print("Encoded dataset:", df_encoded.shape)
print("Scaled dataset:", df_scaled.shape)

print("\nDuplicate rows:", df_clean.duplicated().sum())
print("Missing values in cleaned data:", df_clean.isnull().sum().sum())
print("Missing values in scaled data:", df_scaled.isnull().sum().sum())

print("\nFinal validation complete!")
```
